```python
from pyspark.sql import SparkSession
spark = (
SparkSession.builder
.appName("Economics Analysis")
.master("local[*]")
.getOrCreate())
print("PySpark started successfully")
```

PART A


```python
economic_data = [
("Sri Lanka", 2021, 7.0, 5.1, 3.5,12,18),
("Sri Lanka", 2022, 46.4, 4.7, -7.3,7,15),
("Sri Lanka", 2023, 17.4, 4.7, -2.3,16,21),
("India", 2021, 5.1, 6.2, 9.7,14,18),
("India", 2022, 6.7, 7.3, 7.0,16,24),
("India", 2023, 5.7, 4.8, 8.2,41,36),
("Bangladesh", 2021, 5.5, 5.2, 6.9,14,19),
("Bangladesh", 2022, 7.7, 4.7, 7.1,13,18),
("Bangladesh", 2023, 9.9, 4.5, 5.8,16,13),
("China",2021,4.8,3.2,5.9,58,24)
]
columns = [
"country",
"year",
"inflation_rate",
"unemployment_rate",
"gdp_growth",
"exports",
"imports"
]
df = spark.createDataFrame(economic_data, columns)
df.show()
```

```python
df.show(5)
# FIRST 5 RAWS
```

```python
df.printSchema()
# Print the schema
```

```python
print(df.count())
# number of records
```

PART B

```python
df.select(
"country",
"year",
"inflation_rate").show()
```

```python
from pyspark.sql.functions import col
df.select(
col("country"),col("inflation_rate")).show()
```

```python
high_inflation = df.filter(
col("inflation_rate") >10)
high_inflation.show()
```

```python
negative_gdp = df.filter(
    col("gdp_growth")<0)
negative_gdp.show()
```

```python
lower_unemployment_rate=df.filter(
    col("unemployment_rate")<5)
lower_unemployment_rate.show()
```

```python
df.filter(
col("country") =="Sri Lanka").show()
```

```python
df.filter((col("inflation_rate") >5) &
 (col("gdp_growth") <0)
).show()
```

```python
df.filter(
    (col("country")=="Sri Lanka")|
    (col("country")=="India")
).show()
```

```python
df.orderBy(
    col("inflation_rate").desc()
).show()
```
