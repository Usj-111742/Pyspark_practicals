```python
import wbgapi as wb
import pandas as pd

data = wb.data.DataFrame(
    ['FP.CPI.TOTL.ZG', 'SL.UEM.TOTL.ZS'],
    'LKA',
    time=range(2015,2025)
)


df = pd.DataFrame(data)

print(df)
```

```python
import matplotlib.pyplot as plt

# Transpose the DataFrame to have years as the index
df_plot = df.T

# Set the index name to 'Year'
df_plot.index.name = 'Year'

# Convert index to integer for proper x-axis sorting and labeling
df_plot.index = df_plot.index.str.replace('YR', '').astype(int)

# Plot the data
df_plot.plot(kind='line', figsize=(12, 6))
plt.title('Inflation and Unemployment Rate for Sri Lanka (LKA)')
plt.xlabel('Year')
plt.ylabel('Value')
plt.grid(True)
plt.legend(title='Series')
plt.show()
```

```python
data = wb.data.DataFrame(
    ['FP.CPI.TOTL.ZG', 'SL.UEM.TOTL.ZS', 'NY.GDP.MKTP.CD', 'NY.GDP.MKTP.KD.ZG',
     'NY.GDP.PCAP.CD'],
    'LKA',
    time=range(1975,2025)
)


df = pd.DataFrame(data)

print(df)
```

```python
# Transpose the DataFrame to work with columns
df_transposed = df.T

# Convert index to integer for proper year handling
df_transposed.index = df_transposed.index.str.replace('YR', '').astype(int)

# Create a DataFrame with complete data (no missing values in 'SL.UEM.TOTL.ZS')
df_complete = df_transposed.dropna(subset=['SL.UEM.TOTL.ZS'])

# Create a DataFrame with missing values in 'SL.UEM.TOTL.ZS'
df_missing_uem = df_transposed[df_transposed['SL.UEM.TOTL.ZS'].isnull()]

print("--- DataFrame with complete 'SL.UEM.TOTL.ZS' data ---")
print(df_complete)
print("\n--- DataFrame with missing 'SL.UEM.TOTL.ZS' data ---")
print(df_missing_uem)
```

```python
X = df_complete[['NY.GDP.MKTP.KD.ZG', 'NY.GDP.MKTP.CD']]
y = df_complete['SL.UEM.TOTL.ZS']
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

print(X_scaled)
```

```python
from sklearn.preprocessing import StandardScaler

yscaler = StandardScaler()
y_scaled = yscaler.fit_transform(y.values.reshape(-1, 1))

print(y_scaled)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(X_scaled, y_scaled, test_size=0.2, random_state=42)

print(f"X_train shape: {X_train.shape}")
print(f"X_test shape: {X_test.shape}")
print(f"y_train shape: {y_train.shape}")
print(f"y_test shape: {y_test.shape}")
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

# Initialize the Linear Regression model
model = LinearRegression()

# Fit the model using the training data
model.fit(X_train, y_train)

# Print the coefficients and intercept
print("Coefficients:", model.coef_)
print("Intercept:", model.intercept_)

# Make predictions on the training set
y_train_pred = model.predict(X_train)

# Calculate and print the R-squared score for the training set
r2_train = r2_score(y_train, y_train_pred)
print(f"R-squared score (training set): {r2_train:.4f}")

# Make predictions on the test set
y_test_pred = model.predict(X_test)

# Calculate and print the R-squared score for the test set
r2_test = r2_score(y_test, y_test_pred)
print(f"R-squared score (test set): {r2_test:.4f}")
```
